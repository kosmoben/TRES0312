# Eksempel 5 – Skalarprodukt og vinkel mellom vektorer
**TRES0312 Fysikk — Kapittel 1: Matematiske verktøy**

Skalarproduktet er definert som:
$$\vec{u} \cdot \vec{v} = u_x v_x + u_y v_y = |\vec{u}||\vec{v}|\cos\theta$$

Vi kan bruke dette til å finne vinkelen mellom to vektorer.

```python
import numpy as np

u = np.array([3.0,  4.0])
v = np.array([1.0, -2.0])

dot    = np.dot(u, v)
cos_th = dot / (np.linalg.norm(u) * np.linalg.norm(v))
theta  = np.degrees(np.arccos(cos_th))

print(f"u . v  = {dot}")
print(f"Vinkel = {theta:.1f} grader")
```
