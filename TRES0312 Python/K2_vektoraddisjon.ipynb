Vi legger sammen to vektorer komponentvis:
$$\vec{a} + \vec{b} = (a_x + b_x,\; a_y + b_y)$$

```python
import numpy as np

# Definer vektorene som NumPy-arrays
a = np.array([3.0, -1.0])
b = np.array([-1.0,  4.0])

c = a + b   # addisjon

print(f"a       = {a}")
print(f"b       = {b}")
print(f"a + b   = {c}")
print(f"|a + b| = {np.linalg.norm(c):.2f}")
```
