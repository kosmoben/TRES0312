Vi beregner den kinetiske energien til et legeme:
$$E_k = \frac{1}{2}mv^2$$

```python
# Kinetisk energi: E_k = 0.5 * m * v**2
m = 2.0    # masse i kg
v = 5.0    # fart i m/s

E_k = 0.5 * m * v**2

print(f"Kinetisk energi: {E_k} J")
```

```python
# Potensiell energi: E_p = m * g * h
m = 5.0     # masse i kg
g = 9.81    # tyngdeakselerasjon i m/s^2
h = 10.0    # høyde i m

E_p = m * g * h

print(f"Potensiell energi: {E_p:.2f} J")
```
