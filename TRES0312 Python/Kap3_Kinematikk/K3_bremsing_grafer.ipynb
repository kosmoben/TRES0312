# Aktivitet: Bremsing – akselerasjon, fart og strekning
**TRES0312 Fysikk — Kapittel 3: Kinematikk (Plenumsregning)**

En bil kjører med fart $v_0 = 20\,\mathrm{m/s}$ og bremser med konstant akselerasjon $a = -5\,\mathrm{m/s^2}$ helt til den stopper ($s_0 = 0$).

Siden akselerasjonen er konstant, er bevegelsen gitt ved

$$a(t) = a, \qquad v(t) = v_0 + a\,t, \qquad s(t) = s_0 + v_0\,t + \tfrac{1}{2}a\,t^2$$

Vi regner ut når bilen stopper (når $v(t) = 0$), og tegner alle tre grafene fram til da.

```python
import numpy as np
import matplotlib.pyplot as plt

# -------------------------
# Parametere (fra aktiviteten "Bremsing")
# -------------------------
v0 = 20.0   # utgangsfart [m/s]
s0 = 0.0    # startposisjon [m]
a0 = -5.0   # akselerasjon [m/s^2]
```

```python
# Tiden det tar før bilen stopper (v(t) = 0)
t_stopp = -v0 / a0
print(f"Bilen stopper etter t = {t_stopp:.2f} s")

t = np.linspace(0, t_stopp, 200)
a = np.full_like(t, a0)
v = v0 + a0 * t
s = s0 + v0 * t + 0.5 * a0 * t**2

print(f"Strekning ved stopp: s = {s[-1]:.2f} m")
```

```python
fig, axs = plt.subplots(1, 3, figsize=(12, 4))

axs[0].plot(t, a, color="seagreen", linewidth=2)
axs[0].set_title("Akselerasjon")
axs[0].set_xlabel("Tid [s]")
axs[0].set_ylabel("a [m/s$^2$]")
axs[0].grid(True, alpha=0.3)

axs[1].plot(t, v, color="tomato", linewidth=2)
axs[1].set_title("Fart")
axs[1].set_xlabel("Tid [s]")
axs[1].set_ylabel("v [m/s]")
axs[1].grid(True, alpha=0.3)

axs[2].plot(t, s, color="royalblue", linewidth=2)
axs[2].set_title("Strekning")
axs[2].set_xlabel("Tid [s]")
axs[2].set_ylabel("s [m]")
axs[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
